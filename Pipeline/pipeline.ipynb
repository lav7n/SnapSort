```python
from ultralytics import YOLO
import cv2
import os
import numpy as np

def detect_faces_in_folder(folder_path, confidence_threshold=0.25):
   """Detect faces in all images in a folder."""
   model = YOLO("model.pt")
   image_results = {}

   # Supported image extensions
   valid_extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.gif')

   for filename in os.listdir(folder_path):
       if filename.lower().endswith(valid_extensions):
           image_path = os.path.join(folder_path, filename)
           image = cv2.imread(image_path)

           if image is None:
               print(f"Failed to read image: {filename}")
               continue

           results = model.predict(source=image, conf=confidence_threshold)
           face_boxes = []

           for box in results[0].boxes.xyxy:
               x1, y1, x2, y2 = map(int, box)
               face_boxes.append((x1, y1, x2, y2))
               cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)

           image_results[filename] = face_boxes

   return image_results

folder_path = "/Users/lava/Documents/Code/SnapSort/ML/media"
results = detect_faces_in_folder(folder_path)

# for image_name, faces in results.items():
#     print(f"{image_name}: {len(faces)} faces detected")
#     for i, bbox in enumerate(faces, 1):
#         print(f"  Face {i}: {bbox}")
results

```

```python
import face_recognition
import os

def GetEmbeds(folder):
    embeds = {}
    for img in os.listdir(folder):
        path = os.path.join(folder, img)
        try:
            image = face_recognition.load_image_file(path)
            encodings = face_recognition.face_encodings(image)
            embeds[img] = encodings if encodings else None
        except Exception as e:
            print(f"Error processing {img}: {e}")
    return embeds

# Usage
folder = "/Users/lava/Documents/Code/SnapSort/ML/media"
embeddings = GetEmbeds(folder)
embeddings
```
