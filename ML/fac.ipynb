```python
from ultralytics import YOLO
import cv2

model = YOLO("model.pt")
faces = {}

all_image_paths=["guest1.jpg", "guest2.jpg","guest3.jpg"]

for img_path in all_image_paths:
    img = cv2.imread(img_path)
    results = model.predict(source=img, conf=0.25)
    for i, box in enumerate(results[0].boxes.xyxy):
        x1, y1, x2, y2 = map(int, box)
        face = img[y1:y2, x1:x2]
        face = cv2.resize(face, (224, 224))
        face_id = f"{img_path}_face_{i}"
        faces[face_id] = face
```

```python
for x,y in faces.items():
    cv2.imshow(f"{x}",y)
    cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
import numpy as np
from deepface.DeepFace import represent

embeddings = []

for i,photo in enumerate(faces.values()):  
    result = represent(photo, model_name="VGG-Face", enforce_detection=False,align=True)
    print(result[0])
    embeddings.append(result[0]["embedding"])


centroid_embedding = np.mean(embeddings, axis=0)

print("Centroid Embedding:", centroid_embedding)
```
